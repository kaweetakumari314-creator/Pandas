**18. Combining Dataframes**

Pandas allows us to combine two dataframes in one so that we can analyze it better if both have related info.

we have two way to do this:
- concat() : It is used to join similar datasets vertically.

- merge() : combines DataFrames using a common column.

```python
section_A = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Zain"],
    "Marks": [70, 85, 60, 95, 45]
}

section_B = {
    "Name": ["Mariyum", "Fatima"],
    "Marks": [80, 74]
}

courses = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Zain"],
    "Courses": ["AI", "Web", "AI", "Web", "AI"]
}

df1 = pd.DataFrame(section_A)
df2 = pd.DataFrame(section_B)
df3 = pd.DataFrame(courses)


```

```python
# combining both Dataframes that we just created using concat() method.

pd.concat([df1, df2], ignore_index = True)
```

```python
# combining both Dataframes that we just created using merge() method.
pd.merge(df1, df3, on = "Name")
```
